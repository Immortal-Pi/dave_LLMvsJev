```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()

url = "https://openrouter.ai/api/alpha/decisions"

headers = {
    "Authorization": f"Bearer {os.environ['OPENROUTER_API_KEY']}",
    "Content-Type": "application/json",
}

payload = {
    "model": "typesafe/jev-1.13",
    "state": "there is no enemy around. I have 10 HP and no one will come near my for another 10 mins",
    "questions": {
        "action": {
            "type": "choice",
            "instructions": "What action should the agent take?",
            "criteria": {
                "attack": "Engage the enemy immediately",
                "retreat": "Move away from danger",
                "hide": "Stay nearby but avoid detection",
                "heal": "heal myself"
            }
        }
    }
}

response = requests.post(
    url,
    headers=headers,
    json=payload
)

print(response.status_code)
print(response.text)
```
